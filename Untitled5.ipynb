```python
# 1. Membuat Data Belanjaan

belanjaan = ["Apel", "Susu", "Roti"]
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

print(f"Karakter pertama dari barang pertama ('{belanjaan[0]}'): {belanjaan[0][0]}")
print(f"Karakter terakhir dari barang terakhir ('{belanjaan[-1]}'): {belanjaan[-1][-1]}")
print("-" * 50)

# 2. Hitung Total Harga per Barang
total_apel = harga_jumlah[0] * harga_jumlah[1]
total_susu = harga_jumlah[2] * harga_jumlah[3]
total_roti = harga_jumlah[4] * harga_jumlah[5]

total_belanja = total_apel + total_susu + total_roti

# 3. Cetak Struk Belanja
nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"

print(f"{nama_toko:^50}")
print("=" * 50)
print(f"Pembeli : {nama_pembeli}")
print("-" * 50)
print(f"{'Barang':<15} {'Harga':>10} {'Jumlah':>10} {'Total':>12}")
print("-" * 50)
print(f"{belanjaan[0]:<15} {harga_jumlah[0]:>10} {harga_jumlah[1]:>10} {total_apel:>12}")
print(f"{belanjaan[1]:<15} {harga_jumlah[2]:>10} {harga_jumlah[3]:>10} {total_susu:>12}")
print(f"{belanjaan[2]:<15} {harga_jumlah[4]:>10} {harga_jumlah[5]:>10} {total_roti:>12}")
print("-" * 50)
print(f"{'Total Belanja':<37} {total_belanja:>12}")
print("=" * 50)
print(f"{'Terima kasih sudah belanja! 🙏':^50}")
print("=" * 50)
print()

# 4. Tambahkan Barang Baru ke List
belanjaan.append("Pisang")

print("Isi list belanjaan setelah ditambahkan barang baru:")
print(belanjaan)
```
